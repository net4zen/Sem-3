## Numpy nditer

```python
print(c)
```

```python
for i in c:
    print(i)
```

```python
for i in c:
    for j in i:
        print(j)
```

```python
for i in c:
    for j in i:
        for k in j:
            print(k)
```

```python
for i in np.nditer(c):
    print(i)
```

## Ndenumerate

```python
for i,v in np.ndenumerate(c):
    print("Index-",i,"Value - ",v)
```

## `np.concatenate`
- It is used to join two arrays along the axis

```python
c = np.arange(6).reshape(2,3)
d = np.arange(6,12).reshape(2,3)
print(c)
print(d)
```

```python
np.concatenate((c,d),axis = 0)
```

```python
np.concatenate((c,d),axis = 1)
```

```python
np.concatenate((c,d),axis = 2)  #Throws error - inconsistent dimension
```

## `np.where`
- It returns the indices of elements of an input array where the given condition is satisfied

```python
a = np.array([45,34,55,12,87,34,25,93])
```

```python
np.where(a>50)
```

```python
np.where(a>50,0,a)
```

```python
np.where(a % 2 == 0,0,a)
```

### For a given numpy array = [0,12,45.21,34.99,99.91,32] convert to celsius value sort the array and find the pos at 0

```python
f = np.array([0,12,45.21,34.99,99.91,32])
c = (f-32) *(5/9)
print(c)
sc = np.sort(c)  #sort the array
```

```python
sc
```

```python
np.where(sc==0)
```
