## Module
- A module is a file consisting of a python code. It can have classes, variables and functions.
- There are two types of modules.
      1.Python inbuilt module
      2.User defined modules

## Python inbuilt modules
- These modules comes with python installation. They can be used directly importing them

## Import
- To use the module functions, it has to be imported first. it is imported using import statement

```python
import math
```

```python
print(math.pi)
```

```python
math.cos(2*math.pi)
```

## Alternate way - Not recommended

```python
from math import *
```

```python
cos(pi)
```

## import using aliases

```python
from math import cos as cosine,pi as pieee
```

```python
cosine(pieee)
```

```python
help(cosine)
```

## Userdefined module

```python
%%writefile myfile.py
a=100
a=299
b=400
s="hello world"
```

```python
import myfile
```

```python
print(myfile.a)
```

```python
myfile.s
```

```python
print(mystring)
```

```python
from myfile import a,b,s
```

```python
print(a,b,s)
```

## 496 Write a python program to make a module which contain all the basic functions related to string and import that module in another file and use that fuctions with string given by user.

```python
%%writefile stringmodule.py
def reversethestring(inputstring):
    return inputstring[::-1]
def charactercount(inputstring):
    return len(inputstring) - inputstring.count(" ")
```

```python
from stringmodule import reversethestring,charactercount
```

```python
reversethestring("This is python")
```
