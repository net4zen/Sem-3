### Try Except Else Finally
**SYNTAX**
```python
try:
    statements # this will run first
except:
    statements # this will run if errors occur
else:
    statements # this will run if no error occur
finally:
    statements # this will run no matter which
```

````python
try:
    print("try something here")
except:
    print("error occured")
else:
    print("successfull run")
finally:
    print("this will run no matter what")
````

````python
try:
    print("try something here")
    x=1/0
except:
    print("error occured")
else:
    print("successfull run")
finally:
    print("this will run no matter what")
````

### Constructor Method
- this method is used to instantiate the object using `__init__`

````python
class Employee:
    employeecount = 0
    def __init__(self,name,salary):
        print("object instantiated")
        self.name= name
        print(id(self.name))
        self.salary = salary
        Employee.employeecount+=1
    def display(self):
        return f"The Name is {self.name} and salary is {self.salary}"
    def __str__(self):
        return f"{self.name}"
    def __del__(self):
        print("the object reference is destroyed and garbage collected")
````

````python
emp1 = Employee('python',1000)

emp1.name

id(emp1.name)

emp1.display()

Employee.employeecount

emp2 = Employee('java',2000)

print(emp2.display())

Employee.employeecount
````
