## INHERITANCE

- A class can inherit from other classes
- Inheritance improves code reuse
- Constructor, attributes and methods are inherited from parent to child
- The parent has no access to child class
- Child class ca overrides methods of parent class. It is called method overriding
- super() method is used to invoke parent class methods and constructor


### Single Inheritance

````python
class Human:
    def __init__(self,name):
        self.name = name
    def show(self):
        print(self.name + " is a person")

class Professor(Human):
    pass
````

### `super()` method

- if we write constructor in both parent and child class, the parent class constructor is not available to child class
- `super()` method is used to invoke parent class constructor and method from child class

````python
class Father:
    def __init__(self):
        self.money = 10000
        print("Father class constructor")
    def show(self):
        print("Father class method")

class Son(Father):
    def __init__(self):
        self.money = 100
        self.car = 'BMW'
        super().__init__()
        print("Son class constructor")

    def show(self):
        super().show()
        print("Son class method")
````

````python
s = Son()
````

````python
s.show()
````

````python
s.money
````

````python
s.money = 900
````

````python
s.money
````

## ABSTRACT CLASS

- A Class derived from ABC class which belongs to abc module is known as abstract class in python

```python
from abc import ABC, abstractmethod
```
- Abstract class can have concrete and abstract methods
- Abstract class needs to be extended and implement in child class
- One can not create object of abstract
- It is not necessary to have all methods abstract in abstract class
- The abstract methods of abstract class must be defined in child classes

````python
from abc import ABC, abstractmethod
````

````python
class Defence(ABC): #Abstract Class
    def __init__(self): # Constructor
        self.id = 101  # Attribute

    @abstractmethod #abstract method
    def area(self):
        pass

    def gun(self): # Concrete Method
        print("The gun is AK47")
    
````

````python
class Army(Defence):
    def area(self):
        print("The area is ",self.id)
````

````python
a = Army()
````

````python
a.area()
````

````python
class Navy(Defence):
    def newarea(self):
        print("This will not work")
````

### WAP to create abstract class Shape. it has abstract method calculate_area. Create derived classes Rectangle and Circle and corresponding calculate area methods to find the area

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self,l,b):
        self.l=l
        self.b=b

    def area(self):
        return self.l * self.b

class Circle(Shape):
    def __init__(self,r):
        self.r=r

    def area(self):
        return 3.14*self.r*self.r
````

````python
r = Rectangle(10,5)
````

````python
print("Area of rectangle: ",r.area())
````

````python
c = Circle(20)
print("Area of circle: ",c.area())
````

## Operator Overloading
```
__gt__ -> >
__lt__ -> <
__eq__ -> ==
__add__ -> +
__sub__ -> -
__mul__ -> *
__truediv__ -> /


## Demonstration through Fraction class
```
Fraction1 = 5/8
Fraction2 = 2/3

````python
class Fraction:
    def __init__(self,x,y):
        self.num = x
        self.den = y

    def __str__(self):
        return f"{self.num}/{self.den}"

    def __add__(self,other):
        new_num = self.num*other.den + self.den*other.num
        new_den = self.den * other.den
        return f"{new_num}/{new_den}"

    def __sub__(self,other):
        new_num = self.num*other.den - self.den*other.num
        new_den = self.den * other.den
        return f"{new_num}/{new_den}"

    def __mul__(self,other):
        new_num = self.num * other.num
        new_den = self.den * other.den
        return f"{new_num}/{new_den}"

    def __mul__(self,other):
        new_num = self.num * other.num
        new_den = self.den * other.den
        return f"{new_num}/{new_den}"

    def __truediv__(self,other):
        new_num = self.num * other.den
        new_den = self.den * other.num
        return f"{new_num}/{new_den}"

    def convert_decimal(self):
        return self.num / self.den

    def __gt__(self,other):
        return self.convert_decimal() > other.convert_decimal()
        
    def __lt__(self,other):
        return self.convert_decimal() < other.convert_decimal()
        
    def __eq__(self,other):
        return self.convert_decimal() == other.convert_decimal()
````

````python
f1 = Fraction(5,8)
f2 = Fraction(2,3)
````

````python
print(f1)
````

````python
print(f2)
````

````python
print(f1+f2)
print(f1-f2)
print(f1*f2)
print(f1/f2)
print(f1.convert_decimal())
print(f1<f2)
print(f1>f2)
print(f1==f2)
````

## WAP to create class DATA with value as data member Implement less than, greater than and equality operator for two object of the same class

````python
class Data:

    def __init__(self,value):
        self.m = value

    def __eq__(self,other):
        return self.m == other.m
        
    def __gt__(self,other):
        return self.m > other.m
        
    def __lt__(self,other):
        return self.m < other.m
````

````python
m1 = Data(20)
m2 = Data(30)
print(m1==m2)
print(m1>m2)
print(m1<m2)
````

## METHOD RESOLUTION ORDER IN PYTHON

- It defines an order in which to search for the class to find out which method to use in case of multiple inheritance.
- It is done through C3 linearization algorithm

### Example Demonstration

````python
class F:
    pass
class E:
    pass
class D:
    pass
class C(D,F):
    pass
class B(D,E):
    pass
class A(B,C):
    pass
````

````python
A.mro()
````

### FIRST STEP
```
L[D] = DO
L[E] = EO
L[F] = FO

### SECOND STEP
```
MRO OD C = C + MERGE(L[D],L[F],DF)
= C + MERGE(DO,FO,DF)
= C + D + MERGE(O,FO,F)
= C + D + F + O
= CDFO

### THIRD STEP
```
MRO OF B = B + MERGE(L[D],L[E],DE)
= B + MERGE(DO,EO,DE)
= B + D + MERGE(O,EO,E)
= B + D + E + O
= BDEO

### FOURTH STEP
```
MRO OF A = A + MERGE(L[B],L[C],BC)
= A + MERGE(BDEO,CDFO,BC)
= A + B + MERGE(DEO,CDFO,C)
= A + B + C + MERGE(DEO,DFO)
= A + B + C + D + MERGE(EO,FO)
= A + B + C + D + E + MERGE(O,FO)
= A + B + C + D + E + F + O
= ABCDEFO
