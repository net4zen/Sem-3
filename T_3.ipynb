### Write a python program to search for a string in text files.

````python
%%writefile friends.txt
Friends are crazy, Friends are naughty!
````

````python
def search(filename,query):
    with open(filename) as file:
        content = file.read()
        if query in content:
            print("the string is in the text file")
        else:
            print("the string is not in the text file")

search("third.txt","This")
````

## Q 475
### Write a “pager” program. Your solution should prompt for a filename, and display the text file 25 lines at a time, pausing 
each time to ask the user to enter the word “continue”, in order to show the next 25 lines or enter the word “stop” to 
close the file

````python
def pager(filename):
    with open(filename) as file:
        lines = file.readlines()
        total = len(lines)
        start =0
        end = min(25,total)
        while start<total:
            for i in range(start,end):
                print(lines[i],end="")
            print("---------------------------------------------------")
            user = input("Enter text to continue or stop to stop:")
            print("---------------------------------------------------")
            if user=='stop':
                break;
            start+=25
            end = min(start+25,total)
pager("alice.txt")
        
````
